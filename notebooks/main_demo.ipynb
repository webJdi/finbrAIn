```python
import os
import sys
from pathlib import Path

# Add project root to Python path
project_root = Path().resolve().parent
sys.path.insert(0, str(project_root))

print(f"Project root: {project_root}")
print(f"Python path configured successfully")

from dotenv import load_dotenv
load_dotenv(project_root / ".env")

# Import from src
from src.agents.research_agent import InvestmentResearchAgent
from src.workflows.prompt_chaining import PromptChainingWorkflow
from src.workflows.routing import RoutingWorkflow
from src.workflows.evaluator_optimizer import EvaluatorOptimizerWorkflow
from src.utils.config import FinancialAgentConfig

print("✅ All imports successful!")
```

```python
print("Initializing Financial Agentic AI System...")

config = FinancialAgentConfig()
research_agent = InvestmentResearchAgent()
prompt_chain = PromptChainingWorkflow(config.get_model())
router = RoutingWorkflow(config.get_model())
evaluator = EvaluatorOptimizerWorkflow(config.get_model())

print("🎉 All components initialized successfully!")
```

```python
print("\n" + "="*60)
print("AGENT FUNCTIONS DEMONSTRATION")
print("="*60)

# Function 1: Plan research steps
print(f"\n1. RESEARCH PLANNING for {symbol}")
print("-" * 40)
plan = research_agent.plan_research_steps(symbol)
print("✅ Research plan created successfully!")
print(f"   Steps planned: {len(plan.get('steps', []))}")
for i, step in enumerate(plan.get('steps', []), 1):
    print(f"   {i}. {step}")
```

```python
print(f"\n2. DYNAMIC TOOL USAGE for {symbol}")
print("-" * 40)
analysis = research_agent.get_stock_analysis(symbol)

if 'error' in analysis:
    print(f"❌ Analysis failed: {analysis['error']}")
else:
    print("✅ Stock analysis completed successfully!")
    print(f"   Company: {analysis.get('company_name', 'N/A')}")
    print(f"   Current Price: ${analysis.get('current_analysis', {}).get('price', 'N/A')}")
    print(f"   P/E Ratio: {analysis.get('current_analysis', {}).get('pe_ratio', 'N/A')}")
    print(f"   Recommendation: {analysis.get('recommendation', {}).get('rating', 'N/A')}")
    print(f"   Reasoning: {analysis.get('recommendation', {}).get('reasoning', 'N/A')}")
    
    # Show data sources used
    if 'data_sources_used' in analysis:
        print(f"   Data Sources: {', '.join(analysis['data_sources_used'])}")
```

```python
print(f"\n3. SELF-REFLECTION on Analysis Quality")
print("-" * 40)
reflection = research_agent.self_reflect(analysis)
print("✅ Self-reflection completed!")
print(f"   Quality Score: {reflection.get('quality_assessment', {}).get('overall_score', 'N/A')}/100")
print(f"   Strengths: {len(reflection.get('improvement_suggestions', []))} improvement areas identified")
print(f"   Learning Notes: {len(reflection.get('learning_notes', []))} insights captured")
```

```python
print(f"\n4. LEARNING & MEMORY SYSTEM")
print("-" * 40)

# Test complete workflow
print(f"\n5. COMPLETE RESEARCH WORKFLOW")
print("-" * 40)
full_research = research_agent.run_research(symbol)
print("✅ Complete research workflow executed!")

# Show memory system in action
if 'memory_stats' in full_research:
    stats = full_research['memory_stats']
    print(f"   Memory Entries: {stats.get('total_entries', 0)}")
    print(f"   Research Plans: {stats.get('research_plans', 0)}")
    print(f"   Analyses: {stats.get('analyses', 0)}")
    print(f"   Reflections: {stats.get('reflections', 0)}")

# Show learning insights
if 'learning_insights' in full_research:
    insights = full_research['learning_insights']
    print(f"   Patterns Identified: {len(insights.get('patterns_identified', []))}")
    print(f"   Improvement Areas: {len(insights.get('improvement_areas', []))}")
```

```python
print("\n" + "="*60)
print("WORKFLOW PATTERNS DEMONSTRATION")
print("="*60)

print(f"\n1. PROMPT CHAINING Workflow")
print("-" * 40)
chain_result = prompt_chain.execute_chain(symbol)
print("✅ Prompt chaining workflow completed!")
print(f"   Workflow: {chain_result.get('workflow', 'N/A')}")
print(f"   Stages: {chain_result.get('stages', {})}")
print(f"   News Items Processed: {chain_result.get('stages', {}).get('ingested_news', 0)}")
if 'summary' in chain_result:
    summary = chain_result['summary']
    print(f"   Summary: {summary}")
```

```python
print(f"\n2. ROUTING Workflow")
print("-" * 40)

# Test the routing workflow
routing_test_content = "Apple earnings report shows 15% revenue growth beating expectations"
routing_result = router.auto_route(routing_test_content)

print("✅ Routing workflow completed!")
print(f"   Content: '{routing_test_content[:50]}...'")
print(f"   Routed to: {routing_result.get('analyzer', 'N/A').upper()}")
print(f"   Analysis: {routing_result.get('analysis', 'N/A')}")
```

```python
print(f"\n3. EVALUATOR-OPTIMIZER Workflow")
print("-" * 40)
eval_result = evaluator.execute_workflow(symbol, analysis)
print("✅ Evaluator-optimizer workflow completed!")
print(f"   Iterations: {eval_result.get('iterations', 'N/A')}")
print(f"   Improvement: +{eval_result.get('improvement', 'N/A')}%")
print(f"   Final Grade: {eval_result.get('final_grade', 'N/A')}")
```

```python
print("\n" + "="*60)
print("PROJECT REQUIREMENTS VERIFICATION")
print("="*60)

# Now all variables should be defined since we ran all previous cells
requirements = {
    "AGENT FUNCTIONS (33.8%)": {
        "Research Planning": "plan" in str(full_research).lower() and len(full_research.get('research_plan', {}).get('steps', [])) > 0,
        "Dynamic Tool Usage": "analysis" in full_research and "data_sources_used" in full_research.get('analysis', {}),
        "Self-Reflection": "reflection" in full_research and "quality_assessment" in full_research.get('reflection', {}),
        "Learning/Memory": "memory_stats" in full_research and full_research.get('memory_stats', {}).get('total_entries', 0) > 0
    },
    "WORKFLOW PATTERNS (33.8%)": {
        "Prompt Chaining": chain_result.get('workflow') == 'prompt_chaining',
        "Routing": routing_result.get('workflow') == 'routing',
        "Evaluator-Optimizer": eval_result.get('workflow') == 'evaluator_optimizer'
    }
}

print("\nREQUIREMENTS STATUS:")
print("-" * 40)

total_score = 0
max_score = 0

for category, items in requirements.items():
    print(f"\n{category}:")
    category_score = 0
    for item, met in items.items():
        max_score += 1
        if met:
            total_score += 1
            category_score += 1
            status = "✅"
        else:
            status = "❌"
        print(f"   {status} {item}")
    
    print(f"   Category Score: {category_score}/{len(items)}")

overall_percentage = (total_score / max_score) * 100 if max_score > 0 else 0
print(f"\nOVERALL SCORE: {total_score}/{max_score} ({overall_percentage:.1f}%)")

if overall_percentage >= 80:
    print("🎉 EXCELLENT - All major requirements demonstrated!")
elif overall_percentage >= 60:
    print("✅ GOOD - Most requirements demonstrated")
else:
    print("⚠️  NEEDS IMPROVEMENT - Some requirements missing")

print(f"\nSymbol Analyzed: {symbol}")
print(f"Analysis Completed: {full_research.get('timestamp', 'N/A')}")
```
